DESAFIO 1

Você está ajudando a organizar uma conferência de mulheres na tecnologia e precisa criar crachás com cores diferentes para facilitar o networking. Crie um programa que solicita à participante quantos anos de experiência ela possui e, em seguida, atribui uma categoria:

* **0 a 2 anos:** Crachá Verde (Júnior)
* **3 a 6 anos:** Crachá Amarelo (Pleno)
* **7 anos ou mais:** Crachá Vermelho (Sênior)

```python
#Saudação ao usuário
print ("Olá amiga!")

#Entrada de valor
experiencia = input("Você é da área de tecnologia? ")

#Primeiro bloco de condição para ver se a pessoa é da área de tecnologia
if experiencia.lower() == "sim":

  #Criando variável de anos de experiência
  anos_experiencia = int(input("Quantos anos na área de tecnologia você possui? "))

#Segundo bloco de condição para indicar qual crachá a pessoa deverá usar.
  if anos_experiencia <= 2:
    print("Bem vinda! Você é Júnior (Crachá verde)")

  elif 3 <= anos_experiencia <= 6:
    print("Bem vinda! Você é Pleno (Crachá amarelo)")

  elif anos_experiencia >= 7:
    print("Bem vinda! Você é Sênior (Crachá vermelho)")

#Fecha o bloco de condição com uma condição final caso a pessoa não seja da área da tecnologia
else:
  print("Esse evento é só para mulheres da tecnologia")

```

DESAFIO 2

Durante o bootcamp, as participantes registram o tempo total de estudo em minutos. Para facilitar a análise, você precisa converter esse valor para horas e minutos.

Crie um programa que leia um número inteiro representando o total de minutos estudados e exiba o valor equivalente em horas e minutos.

```python
#Saudação ao usuário
print("Olá amiga!")

#Variável que guarda a quantidade em minutos de estudos
minutos = int(input("Digite aqui em minutos, seu tempo de estudo:  "))

#Transformando os minutos para horas
horas_estudo = minutos / 60
minutos_estudo = minutos % 60

#Retorna o valor da equação e converte para apenas 2 casas decimais
print(f"Você estudou {horas_estudo:.2f} horas e {minutos_estudo} minutos")



```

DESAFIO 3


Você está criando um protótipo para um programa de mentoria que conecta mulheres em início de carreira com mentoras experientes. Crie um programa que faça 3 perguntas de "sim" ou "não" para a usuária e, com base nas respostas, sugira uma trilha de mentoria.

Pergunta 1: "Você gosta de criar relatórios e dashboards?"

Pergunta 2: "Você se interessa por estatística e modelos preditivos?"

Pergunta 3: "Você gosta de trabalhar com a infraestrutura e o fluxo de dados?"


Regras de Sugestão:

Se a resposta para a pergunta 1 for "sim", sugira "Trilha de Analista de BI".

Se a resposta para a pergunta 2 for "sim", sugira "Trilha de Cientista de Dados".

Se a resposta para a pergunta 3 for "sim", sugira "Trilha de Engenheira de Dados".

Se nenhuma for "sim", sugira "Trilha de Fundamentos de Dados".

```python
#Dando valor as varáveis
aula_1 = input("Olá, você gosta de criar relatórios e dashboards? ")
aula_2 = input("Você se interessa por estatística e modelos preditivos? ")
aula_3 = input("Você gosta de trabalhar com a infraestrutura e o fluxo de dados? ")

#variável temporária aula
aula = False

#Blocos de condição
if aula_1.lower() == "sim":
    print("Trilha de Analista de BI")
    aula = True

if aula_2.lower() == "sim":
    print("Trilha de Cientista de Dados")
    aula = True

if aula_3.lower() == "sim":
    print("Trilha de Engenheira de Dados")
    aula = True

if aula == False:
    print("Trilha de Fundamentos de Dados")
```

DESAFIO 4

Crie um pequeno sistema para gerenciar uma lista de artigos sobre tecnologia escritos por mulheres. O sistema deve ser um menu que roda em um loop `while` e oferece as seguintes opções:

1. **Adicionar artigo:** Pede à usuária o título, a autora e uma tag (ex: 'Python', 'Carreira').
2. **Listar todos os artigos:** Exibe todos os artigos cadastrados.
3. **Buscar por autora:** Pede o nome de uma autora e exibe todos os artigos escritos por ela.
4. **Sair**

Use uma lista de dicionários para armazenar os artigos.

```python
# variável que vai armazenar os artigos
artigos = []

# Loop infinito que mantém o menu rodando até o usuário escolher sair
while True:

    # Exibe as opções do menu
    print("\nMENU")
    print("1. Adicionar artigo")
    print("2. Listar todos os artigos")
    print("3. Buscar por autora")
    print("4. Sair")

    # input para escolher uma opção do menu
    opcao = input("Escolha uma opção: ")

    # Início do bloco de condições
    # verifica qual opção o usuário escolheu
    if opcao == "1":

        # solicita o título do artigo
        titulo = input("Digite o título do artigo: ")

        # solicita o nome da autora
        autora = input("Digite o nome da autora: ")

        # parâmetro append para adicionar valores em artigo
        # adiciona um dicionário dentro da lista "artigos"
        # cada artigo terá título e autora
        artigos.append({"titulo": titulo, "autora": autora})

        # mensagem confirmando que o artigo foi cadastrado
        print("Você adicionou um artigo")

    elif opcao == "2":

        # Verifica se a lista de artigos está vazia
        # se estiver vazia, mostra uma mensagem
        if len(artigos) == 0:
            print("Nenhum artigo cadastrado.")

        # caso exista algum artigo cadastrado
        else:
            print("\nLista de artigos:")

            # percorre cada elemento da lista artigos
            for artigo in artigos:

                # mostra o título e a autora de cada artigo
                print(f"Título: {artigo['titulo']} | Autora: {artigo['autora']}")

    elif opcao == "3":

        # input para buscar nome da autora
        busca = input("Digite o nome da autora: ")

        # variável usada para verificar se algum artigo foi encontrado
        encontrados = False

        # percorre todos os artigos cadastrados
        for artigo in artigos:

            # compara o nome da autora digitado com o da lista
            # lower() é usado para ignorar maiúsculas e minúsculas
            if artigo["autora"].lower() == busca.lower():

                # se encontrar, mostra o título do artigo
                print(f"Título: {artigo['titulo']}")

                # muda a variável para True indicando que encontrou resultado
                encontrados = True

        # se nenhum artigo foi encontrado após o loop
        if not encontrados:
            print("Nenhum artigo encontrado para essa autora.")

    elif opcao == "4":

        # mensagem informando que o programa será encerrado
        print("Saindo do programa...")

        # break encerra o loop while e finaliza o programa
        break

    else:
        # caso o usuário digite uma opção que não existe no menu
        print("Opção inválida.")
```
